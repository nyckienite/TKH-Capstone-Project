```python
# Step 1: Normalize column names so "Python" == "python"
df1.columns = df1.columns.str.lower()
df2.columns = df2.columns.str.lower()
```

```python
common_cols = list(set(df1.columns).intersection(set(df2.columns)))
print("Common columns:", common_cols)
```

```python
merged_df = pd.merge(df1, df2, on=common_cols, how="outer")
print(merged_df.shape)

```

```python
print(merged_df.head())

```

```python
print(merged_df.columns)
```

```python
merged_df = merged_df.drop_duplicates()
print(merged_df.shape)
print(merged_df.columns)
```

```python
merged_df[['avg_salary', 'avg salary(k)']].head()
```

```python
merged_df["avg_salary_clean"] = merged_df["avg_salary"].combine_first(merged_df["avg salary(k)"])
```

```python
merged_df[["avg_salary", "avg salary(k)", "avg_salary_clean"]].head()
```

```python
merged_df.drop(columns=["avg_salary", "avg salary(k)"], inplace=True)

merged_df.rename(columns={"avg_salary_clean": "avg_salary"}, inplace=True)
```

```python
merged_df["avg_salary"].head()
```

```python
all_cols = set(merged_df.columns)
common_cols_set = set(common_cols)  # from the merge
extra_cols = list(all_cols - common_cols_set)
print(extra_cols)
```

```python
# Correct skill columns based on dataset
skill_cols = [
    'python', 'sql', 'tensor', 'keras', 'pytorch', 'spark', 
    'hadoop', 'aws', 'tableau', 'big_data', 'bi', 'flink', 
    'scikit', 'mongo', 'google_an'
]

# Columns to keep
columns_to_keep = ['industry', 'job title'] + skill_cols + ['avg_salary']

# Keep only relevant columns
df_clean = merged_df[columns_to_keep].copy()

# Fill NaNs in skill columns with 0
df_clean[skill_cols] = df_clean[skill_cols].fillna(0).astype(int)


print(df_clean.head())
```

```python
df_clean.to_csv("ds_skills_analysis.csv", index=False)
```

```python
df = pd.read_csv("ds_skills_analysis.csv")

# Summary statistics
print(df['avg_salary'].describe())

# Histogram
plt.hist(df['avg_salary'].dropna(), bins=30, edgecolor='black')
plt.title("Distribution of Average Salary")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
# Frequency counts
print(df['industry'].value_counts().head(10))

# Bar plot
df['industry'].value_counts().head(10).plot(kind='bar')
plt.title("Top Industries in Job Postings")
plt.xlabel("Industry")
plt.ylabel("Count")
plt.show()
```

```python
skill_cols = ['python', 'sql', 'tensor', 'keras', 'pytorch', 
              'spark', 'hadoop', 'aws', 'tableau', 'big_data', 
              'bi', 'flink', 'scikit', 'mongo', 'google_an']

skill_counts = df[skill_cols].sum().sort_values(ascending=False)

skill_counts.plot(kind='bar', figsize=(10,6))
plt.title("Frequency of Skills in Job Postings")
plt.xlabel("Skill")
plt.ylabel("Count")
plt.show()
```

```python
skill_counts = df_clean.iloc[:, 2:-1].sum().sort_values(ascending=False)
print(skill_counts)
```

```python
skill_salary = {}
for skill in skill_cols:
    skill_salary[skill] = df_clean.groupby(skill)['avg_salary'].mean()

skill_salary_df = pd.DataFrame(skill_salary).T
skill_salary_df.columns = ['No Skill Mention', 'Skill Mentioned']
skill_salary_df['Difference'] = skill_salary_df['Skill Mentioned'] - skill_salary_df['No Skill Mention']
skill_salary_df.sort_values(by='Difference', ascending=False, inplace=True)
skill_salary_df.head(10)
```

```python
top_skills = skill_salary_df.sort_values(by="Difference", ascending=False).head(10)

top_skills["Difference"].plot(kind="barh", figsize=(8,6))
plt.xlabel("Salary Increase (Skill Mentioned vs Not)")
plt.title("Top 10 Skills Associated with Higher Salaries")
plt.show()
```
