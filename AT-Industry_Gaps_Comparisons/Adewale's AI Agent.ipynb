```python
import os
from google import genai # Using google-genai as an example

# --- Configuration ---
# Use a strong model for complex reasoning tasks
MODEL_NAME = "gemini-2.5-flash" 

class CareerCoachAgent:
    """
    An AI agent for career coaching, skill identification, and gap analysis.
    """
    def __init__(self):
        """Initializes the connection to the LLM API."""
        try:
            # The client automatically picks up the GEMINI_API_KEY environment variable
            self.client = genai.Client() 
        except Exception as e:
            print(f"Error initializing API client. Ensure GEMINI_API_KEY is set.")
            print(f"Details: {e}")
            self.client = None

    def _generate_response(self, prompt: str) -> str:
        """Internal helper to call the LLM."""
        if not self.client:
            return "Error: AI client is not initialized. Please check your API key."

        try:
            print("...Consulting AI Model...")
            # Use the generative model to get a structured response
            response = self.client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt
            )
            return response.text
        except Exception as e:
            return f"An API call error occurred: {e}"

    # --- Core Function 1: Skill Gap and Next Steps ---

    def recommend_next_steps(self, acquired_skills: list[str], target_job: str) -> str:
        """
        Identifies missing skills for a target job and suggests learning resources.
        
        :param acquired_skills: List of skills the learner has.
        :param target_job: The desired job title.
        :return: A detailed plan from the AI.
        """
        skill_list_str = "\n- ".join(acquired_skills)

        prompt = f"""
        You are an expert career coach. A learner wants to acquire the job of a '{target_job}'.
        They currently possess the following skills:
        
        - {skill_list_str}
        
        **Your task is to provide a 3-step action plan:**
        1. **Gap Analysis:** Identify the top 3-5 critical skills they are *missing* for the target job.
        2. **Next Steps:** For each missing skill, suggest the *next practical learning step* (e.g., specific course, project, or concept).
        3. **Direction:** Provide a brief concluding statement to help them identify personal direction and motivation.
        
        Present the response clearly using bullet points and headings.
        """
        return self._generate_response(prompt)

    # --- Core Function 2: Qualification Check and Resume Review ---

    def analyze_candidacy(self, resume_summary: str, job_posting: str) -> str:
        """
        Checks qualification against a job posting and suggests resume improvements.
        
        :param resume_summary: A text summary of the candidate's experience/resume.
        :param job_posting: The full text of the job description.
        :return: A detailed analysis from the AI.
        """
        prompt = f"""
        You are an HR expert and a career coach. Analyze the following candidate summary against the job posting.

        **CANDIDATE SUMMARY:**
        ---
        {resume_summary}
        ---

        **JOB POSTING:**
        ---
        {job_posting}
        ---

        **Your task is to provide a comprehensive analysis:**
        1. **Qualification Status:** State clearly if the candidate is **Well Qualified**, **Qualified**, or **Under Qualified** and why.
        2. **Gap Identification:** Identify 3-5 specific skills, experiences, or keywords missing from the summary that are critical in the job posting.
        3. **Resume Enhancement:** Provide 3 actionable tips for the candidate to immediately improve their resume summary/curriculum to better match this job posting.
        """
        return self._generate_response(prompt)


# --- Example Usage ---

if __name__ == "__main__":
    coach = CareerCoachAgent()

    print("==============================================")
    print("      CAREER COACH AGENT - DEMO MODE          ")
    print("==============================================")
    
    # --- DEMO 1: Next Steps Recommendation ---
    print("\n\n--- DEMO 1: Identifying Next Learning Steps ---")
    
    my_skills = [
        "Python (Beginner)",
        "Data Analysis with Pandas",
        "SQL Queries",
        "Basic Machine Learning Theory",
        "Communication Skills"
    ]
    target = "Senior Data Scientist"
    
    print(f"\n**Target Job:** {target}")
    print(f"**Acquired Skills:** {', '.join(my_skills)}")
    
    recommendation = coach.recommend_next_steps(my_skills, target)
    print("\nAI COACH REPORT:")
    print("-----------------------------------")
    print(recommendation)
    print("-----------------------------------")


    # --- DEMO 2: Qualification Check and Gap Analysis ---
    print("\n\n--- DEMO 2: Analyzing Candidacy Against a Role ---")
    
    candidate_resume = (
        "5 years of experience in customer support. Proficient in Zendesk "
        "and Salesforce. Strong problem-solving abilities and team leadership. "
        "B.A. in Communications. Managed a team of 4 junior agents."
    )
    job_desc = (
        "Seeking a **Product Manager** with 3+ years in a PM role. "
        "Must have experience with **Agile/Scrum methodologies**, "
        "writing **Product Requirements Documents (PRDs)**, and **SQL** proficiency "
        "for data analysis. Experience launching a **SaaS product** is a major plus."
    )
    
    print("\n**Job Posting Snippet:** Product Manager...")
    print(f"**Candidate Summary:** {candidate_resume[:60]}...")
    
    analysis = coach.analyze_candidacy(candidate_resume, job_desc)
    print("\nAI COACH REPORT:")
    print("-----------------------------------")
    print(analysis)
    print("-----------------------------------")
```

```python
import streamlit as st
from career_coach import CareerCoachAgent # Assuming you put the agent in 'career_coach.py'

# 1. Initialize the coach
coach = CareerCoachAgent()

st.title("AI Career Coach")
st.markdown("Enter your details and a target role to get personalized career advice.")

# 2. Collect user input
skills = st.text_area("What skills do you currently possess (one per line)?")
target_job = st.text_input("What is your target job title?")

# 3. Handle the button press
if st.button("Get My Next Steps"):
    if skills and target_job:
        # 4. Call the core AI logic
        skill_list = [s.strip() for s in skills.split('\n') if s.strip()]
        with st.spinner('Analyzing your career path...'):
            recommendation = coach.recommend_next_steps(skill_list, target_job)

        # 5. Display the output
        st.subheader("Your Personalized Career Plan")
        st.write(recommendation)
    else:
        st.error("Please fill in both your skills and the target job.")
```
